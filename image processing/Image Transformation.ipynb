## 2x2 Linear Transformation of Points

A 2D linear transformation maps a point $(x, y)$ to a new point $(x', y')$ using a $2 \times 2$ transformation matrix $A$:

$$
\begin{bmatrix} x' \\ y' \end{bmatrix} = \begin{bmatrix} a & b \\ c & d \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix}
$$

Expanding this matrix multiplication gives the coordinate equations:

* $x' = ax + by$
* $y' = cx + dy$

### Common 2x2 Transformation Matrices

| Transformation | Matrix $A$ | Effect on $(x, y)$ |
| :--- | :---: | :--- |
| **Scaling** | $\begin{bmatrix} s_x & 0 \\ 0 & s_y \end{bmatrix}$ | Stretches by $s_x$ horizontally and $s_y$ vertically |
| **Rotation** (angle $\theta$) | $\begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}$ | Rotates counterclockwise around the origin by $\theta$ |
| **Horizontal Shear** | $\begin{bmatrix} 1 & k \\ 0 & 1 \end{bmatrix}$ | Shifts $x$ proportionally to $y$ ($x' = x + ky$) |
| **Reflection (x-axis)** | $\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}$ | Flips the point vertically across the x-axis |

### Transforming Multiple Points

To transform $n$ points simultaneously, arrange them as columns in a $2 \times n$ matrix $P$:

$$
P' = A P = \begin{bmatrix} a & b \\ c & d \end{bmatrix} \begin{bmatrix} x_1 & x_2 & \dots & x_n \\ y_1 & y_2 & \dots & y_n \end{bmatrix}
$$

```python
def transform_points(matrix, points):
    """Apply a 2x2 matrix [[a, b], [c, d]] to a list of (x, y) tuples."""
    a, b = matrix[0]
    c, d = matrix[1]
    return [(a * x + b * y, c * x + d * y) for x, y in points]


A = [
    [2.0, -1.0],
    [0.5,  1.5],
]
points = [(1.0, 2.0), (3.0, -1.0), (0.0, 4.0)]

result = transform_points(A, points)
for orig, trans in zip(points, result):
    print(f"{orig} -> {trans}")
```

```python
import numpy as np

# 2x2 transformation matrix A
A = np.array([
    [2.0, -1.0],
    [0.5,  1.5]
])

# Input points as rows: (x, y)
points = np.array([
    [1.0,  2.0],
    [3.0, -1.0],
    [0.0,  4.0]
])

# Apply A to each point: (A @ points.T).T yields shape (n, 2)
transformed_points = (A @ points.T).T

for orig, trans in zip(points, transformed_points):
    print(f"({orig[0]}, {orig[1]}) -> ({trans[0]}, {trans[1]})")
```

## Scaling from $(x_1, y_1)$ to $(x_2, y_2)$

Scaling changes the size of a point or shape relative to a chosen **anchor (pivot) point** using horizontal and vertical scale factors $(s_x, s_y)$.

### 1. Scaling Relative to the Origin $(0, 0)$

If the origin $(0, 0)$ is the fixed center of scaling, the scale factors from $(x_1, y_1)$ to $(x_2, y_2)$ are:

$$
s_x = \frac{x_2}{x_1} \quad (\text{for } x_1 \neq 0), \qquad s_y = \frac{y_2}{y_1} \quad (\text{for } y_1 \neq 0)
$$

In matrix form, a $2 \times 2$ scaling matrix $S$ maps $(x_1, y_1)$ to $(x_2, y_2)$:

$$
\begin{bmatrix} x_2 \\ y_2 \end{bmatrix} = \begin{bmatrix} s_x & 0 \\ 0 & s_y \end{bmatrix} \begin{bmatrix} x_1 \\ y_1 \end{bmatrix}
$$

* **Uniform scaling:** $s_x = s_y = s$ (preserves aspect ratio).
* **Non-uniform scaling:** $s_x \neq s_y$ (stretches or squashes along one axis).
* **Magnitude vs. direction:** $|s| > 1$ enlarges, $0 < |s| < 1$ shrinks, and $s < 0$ flips across that axis.

---

### 2. Scaling Relative to a Pivot / Anchor Point $(p_x, p_y)$

A pure $2 \times 2$ matrix always keeps $(0, 0)$ fixed. To scale around an arbitrary anchor $(p_x, p_y)$:

1. **Translate** the point so $(p_x, p_y)$ moves to the origin: $(x - p_x,\; y - p_y)$.
2. **Scale** using $(s_x, s_y)$: $(s_x(x - p_x),\; s_y(y - p_y))$.
3. **Translate back** by adding $(p_x, p_y)$:

$$
\begin{aligned}
x' &= p_x + s_x(x - p_x) \\
y' &= p_y + s_y(y - p_y)
\end{aligned}
$$

If you know an initial point $(x_1, y_1)$, a target point $(x_2, y_2)$, and a pivot $(p_x, p_y)$ with $x_1 \neq p_x$ and $y_1 \neq p_y$, the implied scale factors are:

$$
s_x = \frac{x_2 - p_x}{x_1 - p_x}, \qquad s_y = \frac{y_2 - p_y}{y_1 - p_y}
$$

```python
import numpy as np

# Initial point (x1, y1) and target point (x2, y2) relative to origin (0, 0)
x1, y1 = 2.0, 5.0
x2, y2 = 6.0, 2.5

# Compute scale factors (assuming x1 != 0 and y1 != 0)
sx = x2 / x1
sy = y2 / y1
print(f"Scale factors: sx = {sx}, sy = {sy}")

# 2x2 origin-scaling matrix
S = np.array([
    [sx,  0.0],
    [0.0, sy]
])

# Additional points to scale from the origin: rows are (x, y)
points = np.array([
    [2.0,  5.0],
    [4.0,  8.0],
    [-1.0, 6.0]
])

scaled_points = (S @ points.T).T

for orig, scaled in zip(points, scaled_points):
    print(f"({orig[0]}, {orig[1]}) -> ({scaled[0]}, {scaled[1]})")
```

```python
def origin_scale_factors(x1, y1, x2, y2):
    """Return horizontal and vertical scale factors from (x1, y1) to (x2, y2) about (0, 0)."""
    if x1 == 0 or y1 == 0:
        raise ValueError("x1 and y1 must be non-zero to compute origin scale factors.")
    return x2 / x1, y2 / y1


def scale_from_origin(points, sx, sy):
    """Scale a list of (x, y) points relative to the origin (0, 0)."""
    return [(sx * x, sy * y) for x, y in points]


x1, y1 = 2.0, 5.0
x2, y2 = 6.0, 2.5

sx, sy = origin_scale_factors(x1, y1, x2, y2)
print(f"Scale factors: sx = {sx}, sy = {sy}")

points = [(2.0, 5.0), (4.0, 8.0), (-1.0, 6.0)]
scaled = scale_from_origin(points, sx, sy)

for orig, new_pt in zip(points, scaled):
    print(f"{orig} -> {new_pt}")
```

## Rotation from $(x_1, y_1)$ to $(x_2, y_2)$

A 2D rotation around the origin $(0, 0)$ turns a point $(x, y)$ by an angle $\theta$ (counterclockwise for $\theta > 0$) to produce a new point $(x', y')$:

$$
\begin{bmatrix} x' \\ y' \end{bmatrix} = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix}
$$

Expanding this gives the coordinate equations:

$$
\begin{aligned}
x' &= x\cos\theta - y\sin\theta \\
y' &= x\sin\theta + y\cos\theta
\end{aligned}
$$

---

### 1. Finding the Rotation Angle $\theta$ from $(x_1, y_1)$ to $(x_2, y_2)$

Suppose $(x_1, y_1)$ and $(x_2, y_2)$ are non-zero points around the origin $(0, 0)$. Using polar angles $\phi_1 = \operatorname{atan2}(y_1, x_1)$ and $\phi_2 = \operatorname{atan2}(y_2, x_2)$:

$$
\theta = \phi_2 - \phi_1 = \operatorname{atan2}(y_2, x_2) - \operatorname{atan2}(y_1, x_1)
$$

Equivalently, using the dot product and 2D cross product (determinant) of $\mathbf{v}_1 = (x_1, y_1)$ and $\mathbf{v}_2 = (x_2, y_2)$:

$$
\theta = \operatorname{atan2}(x_1 y_2 - y_1 x_2,\; x_1 x_2 + y_1 y_2)
$$

* **Why `atan2`?** It handles all four quadrants cleanly and avoids division-by-zero errors along the axes.
* **Pure rotation vs. similarity:** A pure $2 \times 2$ rotation matrix preserves vector length ($\|\mathbf{v}_2\| = \|\mathbf{v}_1\|$). If $\|\mathbf{v}_2\| \neq \|\mathbf{v}_1\|$, rotating $(x_1, y_1)$ by $\theta$ aligns its direction with $(x_2, y_2)$ without changing its radius unless a scale factor is also applied.

---

### 2. Applying the Rotation to Other Points

Once $\theta$ is known, build the $2 \times 2$ rotation matrix $R(\theta)$ and apply it to any point $(x, y)$:

$$
R(\theta) = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}, \qquad \begin{bmatrix} x' \\ y' \end{bmatrix} = R(\theta) \begin{bmatrix} x \\ y \end{bmatrix}
$$

## Step-by-Step Derivation of 2D Rotation from $P_1(x_1, y_1)$ to $P_2(x_2, y_2)$

### Step 1: Express $P_1(x_1, y_1)$ in Polar Coordinates
Point $P_1$ lies at a radius $r$ from the origin $(0, 0)$ at an angle $\varphi$ from the positive $x$-axis:

$$
\begin{aligned}
x_1 &= r\cos\varphi \\
y_1 &= r\sin\varphi
\end{aligned}
$$

### Step 2: Rotate by Angle $\theta$ to Reach $P_2(x_2, y_2)$
Rotating $P_1$ counterclockwise around the origin by angle $\theta$ keeps the radius $r$ unchanged and increases the angle from the positive $x$-axis to $\varphi + \theta$:

$$
\begin{aligned}
x_2 &= r\cos(\varphi + \theta) \\
y_2 &= r\sin(\varphi + \theta)
\end{aligned}
$$

### Step 3: Expand Using Angle-Addition Identities
Apply $\cos(\varphi + \theta) = \cos\varphi\cos\theta - \sin\varphi\sin\theta$ and $\sin(\varphi + \theta) = \sin\varphi\cos\theta + \cos\varphi\sin\theta$:

$$
\begin{aligned}
x_2 &= r\cos\varphi\cos\theta - r\sin\varphi\sin\theta \\
y_2 &= r\cos\varphi\sin\theta + r\sin\varphi\cos\theta
\end{aligned}
$$

### Step 4: Substitute $x_1$ and $y_1$ Back In
Replace $r\cos\varphi$ with $x_1$ and $r\sin\varphi$ with $y_1$:

$$
\begin{aligned}
x_2 &= x_1\cos\theta - y_1\sin\theta \\
y_2 &= x_1\sin\theta + y_1\cos\theta
\end{aligned}
$$

In $2 \times 2$ matrix form:

$$
\begin{bmatrix} x_2 \\ y_2 \end{bmatrix} = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix} \begin{bmatrix} x_1 \\ y_1 \end{bmatrix}
$$

```python
import math
import numpy as np

# Given point P1(x1, y1) and counterclockwise rotation angle theta (in degrees)
x1, y1 = 4.0, 3.0
theta_deg = 30.0
theta = math.radians(theta_deg)

# Step 1: Polar coordinates of P1 -> x1 = r * cos(phi), y1 = r * sin(phi)
r = math.hypot(x1, y1)
phi = math.atan2(y1, x1)

print(f"Step 1 (Polar P1): r = {r:.4f}, phi = {math.degrees(phi):.2f} deg")
print(f"  Check x1 = r*cos(phi) = {r * math.cos(phi):.4f}, y1 = r*sin(phi) = {r * math.sin(phi):.4f}")

# Step 2: Rotate by theta -> x2 = r * cos(phi + theta), y2 = r * sin(phi + theta)
x2_polar = r * math.cos(phi + theta)
y2_polar = r * math.sin(phi + theta)
print(f"Step 2 (Angle addition): P2 = ({x2_polar:.4f}, {y2_polar:.4f})")

# Step 3: Expanded sum identities -> r*cos(phi)*cos(theta) - r*sin(phi)*sin(theta), etc.
x2_expanded = r * math.cos(phi) * math.cos(theta) - r * math.sin(phi) * math.sin(theta)
y2_expanded = r * math.cos(phi) * math.sin(theta) + r * math.sin(phi) * math.cos(theta)
print(f"Step 3 (Expanded terms): P2 = ({x2_expanded:.4f}, {y2_expanded:.4f})")

# Step 4: Substitute x1, y1 & apply 2x2 rotation matrix
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])
p1 = np.array([x1, y1])
p2_matrix = R @ p1
print(f"Step 4 (2x2 Matrix):    P2 = ({p2_matrix[0]:.4f}, {p2_matrix[1]:.4f})")
```

## Skew (Shear) from $(x_1, y_1)$ to $(x_2, y_2)$

A 2D **skew** (or **shear**) transformation shifts one coordinate proportionally to the other while keeping the opposite axis parallel. Relative to the origin $(0, 0)$, the two elementary $2 \times 2$ shear matrices are:

* **Horizontal shear (along $x$):** shifts $x$ in proportion to $y$ while keeping $y$ unchanged.
* **Vertical shear (along $y$):** shifts $y$ in proportion to $x$ while keeping $x$ unchanged.

---

### 1. Horizontal Skew (Shear along $x$)

In matrix form:

$$
\begin{bmatrix} x' \\ y' \end{bmatrix} = \begin{bmatrix} 1 & k_x \\ 0 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix}
$$

Expanding this gives the coordinate equations:

$$
\begin{aligned}
x' &= x + k_x y \\
y' &= y
\end{aligned}
$$

If a point $(x_1, y_1)$ with $y_1 \neq 0$ moves to $(x_2, y_2)$ under a pure horizontal skew ($y_2 = y_1$), the horizontal shear factor $k_x$ is:

$$
k_x = \frac{x_2 - x_1}{y_1}
$$

---

### 2. Vertical Skew (Shear along $y$)

In matrix form:

$$
\begin{bmatrix} x' \\ y' \end{bmatrix} = \begin{bmatrix} 1 & 0 \\ k_y & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix}
$$

Expanding this gives the coordinate equations:

$$
\begin{aligned}
x' &= x \\
y' &= y + k_y x
\end{aligned}
$$

If a point $(x_1, y_1)$ with $x_1 \neq 0$ moves to $(x_2, y_2)$ under a pure vertical skew ($x_2 = x_1$), the vertical shear factor $k_y$ is:

$$
k_y = \frac{y_2 - y_1}{x_1}
$$

---

### 3. Combined Skew

Applying both horizontal and vertical shear factors $(k_x, k_y)$ gives:

$$
\begin{bmatrix} x' \\ y' \end{bmatrix} = \begin{bmatrix} 1 & k_x \\ k_y & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} \implies \begin{aligned} x' &= x + k_x y \\ y' &= y + k_y x \end{aligned}
$$

* Note: Unlike pure horizontal or pure vertical shear, the combined matrix $\begin{bmatrix} 1 & k_x \\ k_y & 1 \end{bmatrix}$ is not a simple single-axis shear because $x'$ depends on $y$ and $y'$ depends on $x$ simultaneously (its determinant is $1 - k_x k_y$).

```python
import numpy as np


def apply_matrix(matrix, points):
    """Apply a 2x2 matrix to an (n, 2) array of (x, y) points."""
    return (matrix @ points.T).T


# Sample points to transform: rows are (x, y)
sample_points = np.array([
    [1.0,  2.0],
    [3.0,  4.0],
    [-2.0, 2.0],
])

# ---------------------------------------------------------
# 1. Horizontal Skew: (x1, y1) = (2, 4) -> (x2, y2) = (10, 4)
# ---------------------------------------------------------
x1_h, y1_h = 2.0, 4.0
x2_h, y2_h = 10.0, 4.0
kx = (x2_h - x1_h) / y1_h

H_horiz = np.array([
    [1.0, kx],
    [0.0, 1.0]
])

print(f"--- 1. Horizontal Skew (kx = {kx}) ---")
for orig, out in zip(sample_points, apply_matrix(H_horiz, sample_points)):
    print(f"({orig[0]}, {orig[1]}) -> ({out[0]}, {out[1]})")

# ---------------------------------------------------------
# 2. Vertical Skew: (x1, y1) = (5, 1) -> (x2, y2) = (5, -4)
# ---------------------------------------------------------
x1_v, y1_v = 5.0, 1.0
x2_v, y2_v = 5.0, -4.0
ky = (y2_v - y1_v) / x1_v

H_vert = np.array([
    [1.0, 0.0],
    [ky,  1.0]
])

print(f"\n--- 2. Vertical Skew (ky = {ky}) ---")
for orig, out in zip(sample_points, apply_matrix(H_vert, sample_points)):
    print(f"({orig[0]}, {orig[1]}) -> ({out[0]}, {out[1]})")

# ---------------------------------------------------------
# 3. Combined Skew: kx_comb = 1.5, ky_comb = 0.5
# ---------------------------------------------------------
kx_comb = 1.5
ky_comb = 0.5

H_comb = np.array([
    [1.0,     kx_comb],
    [ky_comb, 1.0]
])

print(f"\n--- 3. Combined Skew (kx = {kx_comb}, ky = {ky_comb}) ---")
for orig, out in zip(sample_points, apply_matrix(H_comb, sample_points)):
    print(f"({orig[0]}, {orig[1]}) -> ({out[0]}, {out[1]})")
```

## Mirror (Reflection) from $(x_1, y_1)$ to $(x_2, y_2)$

A 2D **mirror** (or **reflection**) flips a point across a mirror line (axis) through the origin $(0, 0)$ using a $2 \times 2$ orthogonal matrix with determinant $-1$:

$$
\begin{bmatrix} x' \\ y' \end{bmatrix} = M \begin{bmatrix} x \\ y \end{bmatrix}, \qquad \det(M) = -1
$$

Unlike scaling, rotation, or skew—which map a single point $(x_1, y_1)$ to $(x_2, y_2)$—a reflection depends on the **mirror line** passing through the midpoint $\left(\frac{x_1 + x_2}{2},\; \frac{y_1 + y_2}{2}\right)$ and perpendicular to the segment from $(x_1, y_1)$ to $(x_2, y_2)$.

---

### 1. Common Reflections Through the Origin $(0, 0)$

| Mirror Line | Matrix $M$ | Coordinate Equations |
| :--- | :---: | :--- |
| **Across $x$-axis** ($y = 0$) | $\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}$ | $x' = x,\quad y' = -y$ |
| **Across $y$-axis** ($x = 0$) | $\begin{bmatrix} -1 & 0 \\ 0 & 1 \end{bmatrix}$ | $x' = -x,\quad y' = y$ |
| **Across line $y = x$** | $\begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}$ | $x' = y,\quad y' = x$ |
| **Across line $y = -x$** | $\begin{bmatrix} 0 & -1 \\ -1 & 0 \end{bmatrix}$ | $x' = -y,\quad y' = -x$ |

* **Across $x$-axis:** $(x_1, y_1) \to (x_1, -y_1)$
* **Across $y$-axis:** $(x_1, y_1) \to (-x_1, y_1)$
* **Across $y = x$:** $(x_1, y_1) \to (y_1, x_1)$

---

### 2. General Reflection Across a Line at Angle $\alpha$ Through the Origin

If the mirror line passes through $(0, 0)$ at an angle $\alpha$ from the positive $x$-axis (with unit direction vector $\mathbf{u} = (\cos\alpha, \sin\alpha)$), the $2 \times 2$ reflection matrix is:

$$
M(\alpha) = \begin{bmatrix} \cos 2\alpha & \sin 2\alpha \\ \sin 2\alpha & -\cos 2\alpha \end{bmatrix}
$$

Given $(x_1, y_1)$ and its reflection $(x_2, y_2)$ across a line through $(0, 0)$:
1. The midpoint is $\mathbf{m} = \left(\frac{x_1 + x_2}{2},\; \frac{y_1 + y_2}{2}\right)$, which lies on the mirror line.
2. The mirror line angle $\alpha$ (for a non-zero midpoint) is:

$$
\alpha = \operatorname{atan2}\!\left(\frac{y_1 + y_2}{2},\; \frac{x_1 + x_2}{2}\right) = \operatorname{atan2}(y_1 + y_2,\; x_1 + x_2)
$$

3. Substituting $\alpha$ into $M(\alpha)$ gives the $2 \times 2$ matrix that mirrors $(x_1, y_1)$ to $(x_2, y_2)$ and reflects any other point $(x, y)$ across that same origin-passing line.

```python
import math
import numpy as np


def apply_matrix(matrix, points):
    """Apply a 2x2 matrix to an (n, 2) array of (x, y) points."""
    return (matrix @ points.T).T


def reflection_matrix_from_points(x1, y1, x2, y2):
    """Compute 2x2 origin-passing reflection matrix mapping (x1, y1) to (x2, y2)."""
    sx = x1 + x2
    sy = y1 + y2
    if math.hypot(sx, sy) == 0:
        raise ValueError("Midpoint is at the origin; mirror line angle is undefined.")

    alpha = math.atan2(sy, sx)
    c2 = math.cos(2.0 * alpha)
    s2 = math.sin(2.0 * alpha)

    M = np.array([
        [c2,  s2],
        [s2, -c2]
    ])
    return M, alpha


sample_points = np.array([
    [3.0,  4.0],
    [-2.0, 5.0],
    [1.0, -3.0],
])

# 1. Reflection across x-axis (y = 0)
M_x = np.array([
    [1.0,  0.0],
    [0.0, -1.0]
])
print("--- 1. Across x-axis ---")
for orig, out in zip(sample_points, apply_matrix(M_x, sample_points)):
    print(f"({orig[0]}, {orig[1]}) -> ({out[0]}, {out[1]})")

# 2. Reflection across y-axis (x = 0)
M_y = np.array([
    [-1.0, 0.0],
    [ 0.0, 1.0]
])
print("\n--- 2. Across y-axis ---")
for orig, out in zip(sample_points, apply_matrix(M_y, sample_points)):
    print(f"({orig[0]}, {orig[1]}) -> ({out[0]}, {out[1]})")

# 3. Reflection across line y = x
M_yx = np.array([
    [0.0, 1.0],
    [1.0, 0.0]
])
print("\n--- 3. Across line y = x ---")
for orig, out in zip(sample_points, apply_matrix(M_yx, sample_points)):
    print(f"({orig[0]}, {orig[1]}) -> ({out[0]}, {out[1]})")

# 4. General mirror line from (x1, y1) = (4, 2) to (x2, y2) = (2, 4)
x1, y1 = 4.0, 2.0
x2, y2 = 2.0, 4.0
M_gen, alpha = reflection_matrix_from_points(x1, y1, x2, y2)

print(f"\n--- 4. General Reflection: ({x1}, {y1}) -> ({x2}, {y2}) ---")
print(f"Mirror angle alpha = {math.degrees(alpha):.2f} deg")
print("Matrix M(alpha):\n", np.round(M_gen, 4))

p1_mapped = M_gen @ np.array([x1, y1])
print(f"Check ({x1}, {y1}) -> ({p1_mapped[0]:.4f}, {p1_mapped[1]:.4f})")

for orig, out in zip(sample_points, apply_matrix(M_gen, sample_points)):
    print(f"({orig[0]}, {orig[1]}) -> ({out[0]:.4f}, {out[1]:.4f})")
```

## Cropping from $(x_1, y_1)$ to $(x_2, y_2)$

Unlike **scaling**, **rotation**, **skew**, or **mirroring**, **cropping** is **not** a $2 \times 2$ linear transformation. Instead, it defines a **bounding box (region of interest / ROI)** using two corner points—typically top-left $(x_1, y_1)$ and bottom-right $(x_2, y_2)$—and keeps only the points, pixels, or region inside that rectangle.

---

### 1. Bounding Box Definition

Given corner points $(x_1, y_1)$ and $(x_2, y_2)$, normalize the bounds so they work regardless of which corner is larger:

$$
\begin{aligned}
x_{\min} &= \min(x_1, x_2), \qquad x_{\max} = \max(x_1, x_2) \\
y_{\min} &= \min(y_1, y_2), \qquad y_{\max} = \max(y_1, y_2)
\end{aligned}
$$

The width and height of the cropped region are:

$$
w = x_{\max} - x_{\min}, \qquad h = y_{\max} - y_{\min}
$$

---

### 2. Cropping 2D Points

A point $(x, y)$ lies inside (or on the boundary of) the crop box if and only if:

$$
x_{\min} \le x \le x_{\max} \qquad \text{and} \qquad y_{\min} \le y \le y_{\max}
$$

* Points satisfying both inequalities are **kept**.
* Points outside $[x_{\min}, x_{\max}] \times [y_{\min}, y_{\max}]$ are **discarded** (or clipped to the box edges if clamping).

---

### 3. Cropping a Grid / Image

For a 2D array or image indexed by row $r$ (along $y$) and column $c$ (along $x$):

* **Column (horizontal / $x$) slice:** $c \in [x_{\min}, x_{\max})$ (half-open interval $[x_{\min}, x_{\max})$ has width $x_{\max} - x_{\min}$).
* **Row (vertical / $y$) slice:** $r \in [y_{\min}, y_{\max})$ (half-open interval $[y_{\min}, y_{\max})$ has height $y_{\max} - y_{\min}$).

In slice notation, the cropped sub-grid is:

$$
\text{crop} = \text{grid}[y_{\min} : y_{\max},\; x_{\min} : x_{\max}]
$$

* **Why not a $2 \times 2$ matrix?** Cropping filters/slices data based on **inequalities** ($\min \le \cdot \le \max$) rather than multiplying coordinates by a linear matrix; a $2 \times 2$ matrix always maps $(0, 0) \to (0, 0)$ and cannot select a subset of points or slice a sub-array.

```python
import numpy as np

# Corner points defining the crop box
x1, y1 = 2, 1
x2, y2 = 6, 4

# 1. Normalize bounding box coordinates
x_min, x_max = min(x1, x2), max(x1, x2)
y_min, y_max = min(y1, y2), max(y1, y2)
width = x_max - x_min
height = y_max - y_min

print(f"Bounds: x in [{x_min}, {x_max}], y in [{y_min}, {y_max}]")
print(f"Box size: width = {width}, height = {height}")

# 2. Crop 2D points using inclusive bounds: x_min <= x <= x_max and y_min <= y <= y_max
points = np.array([
    [2, 1],  # inside (on boundary)
    [4, 3],  # inside
    [6, 4],  # inside (on boundary)
    [1, 2],  # outside (x < x_min)
    [7, 3],  # outside (x > x_max)
    [3, 5],  # outside (y > y_max)
])

mask = (
    (points[:, 0] >= x_min) & (points[:, 0] <= x_max) &
    (points[:, 1] >= y_min) & (points[:, 1] <= y_max)
)
cropped_points = points[mask]

print("\n--- Cropped 2D Points ---")
for pt in cropped_points:
    print(f"({pt[0]}, {pt[1]})")

# 3. Crop a 2D grid (row = y, col = x) using slice grid[y_min:y_max, x_min:x_max]
grid = np.arange(8 * 6).reshape(8, 6)  # shape (8, 6): 8 rows (y=0..7), 6 cols (x=0..5)

# Using box x in [1, 5] and y in [2, 5] so x_max <= 6 and y_max <= 8
gx_min, gx_max = 1, 5
gy_min, gy_max = 2, 5
cropped_grid = grid[gy_min:gy_max, gx_min:gx_max]

print(f"\n--- Cropped Grid [y={gy_min}:{gy_max}, x={gx_min}:{gx_max}] (shape {cropped_grid.shape}) ---")
print(cropped_grid)
```
