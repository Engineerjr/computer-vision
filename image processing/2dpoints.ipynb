```python
import numpy as np
import matplotlib.pyplot as plt

# Define 2D points (pixel coordinates)
points = np.array([
    [100, 150],
    [200, 250],
    [300, 100],
    [400, 300],
    [150, 400]
])

# Extract x and y coordinates
x = points[:, 0]
y = points[:, 1]

# Plot points
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='red', s=80)

# Label each point
for i, (xi, yi) in enumerate(points):
    plt.text(xi + 5, yi + 5, f'P{i+1}')

# Configure image coordinate system
plt.gca().invert_yaxis()
plt.xlabel('X (pixels)')
plt.ylabel('Y (pixels)')
plt.title('Representation of 2D Points in an Image')
plt.grid(True)

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Create a blank image
image = 255 * np.ones((500, 600, 3), dtype=np.uint8)

# Define 2D points
points = [
    (100, 150),
    (200, 250),
    (300, 100),
    (400, 300),
    (150, 400)
]

# Draw points and labels
for i, point in enumerate(points):
    cv2.circle(image, point, 5, (0, 0, 255), -1)
    cv2.putText(image, f'P{i+1}',
                (point[0]+10, point[1]),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5, (255, 0, 0), 1)

# Display image
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('2D Pixel Coordinates')
plt.xlabel('X (pixels)')
plt.ylabel('Y (pixels)')
plt.show()
```

## 2D LINES 


```python
import numpy as np
import matplotlib.pyplot as plt

# Define two 2D points in homogeneous coordinates
P1 = np.array([50, 50, 1])
P2 = np.array([150, 150, 1])

# Calculate line using cross product
L = np.cross(P1, P2)

a, b, c = L

print("Homogeneous line coordinates:")
print(L)

print(f"Line equation: {a}x + {b}y + {c} = 0")

# Generate line
x = np.linspace(0, 200, 500)
y = -(a*x + c)/b

# Plot
plt.figure(figsize=(8, 6))

plt.plot(x, y, label='Line')

plt.scatter([50, 150], [50, 150],
            color='red', label='Points')

plt.gca().invert_yaxis()

plt.xlabel('X (pixels)')
plt.ylabel('Y (pixels)')
plt.title('Line Through Two 2D Points')

plt.grid(True)
plt.legend()
plt.show()
```

## Python code to represent and plot 3D points

```python
import numpy as np
import matplotlib.pyplot as plt

# Define 3D points (x, y, z)
points = np.array([
    [2, 3, 5],
    [4, 2, 6],
    [3, 5, 2],
    [6, 4, 3],
    [1, 2, 4]
])

# Extract coordinates
x = points[:, 0]
y = points[:, 1]
z = points[:, 2]

# Create 3D plot
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot points
ax.scatter(x, y, z, color='red', s=80)

# Label points
for i, (xi, yi, zi) in enumerate(points):
    ax.text(xi, yi, zi, f'P{i+1}')

# Axis labels
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

ax.set_title('Representation of 3D Points')

plt.show()
```

# 3D Planes – Homogeneous Coordinates in Computer Vision

```python
import numpy as np
import matplotlib.pyplot as plt

# Define plane coefficients
a, b, c, d = 2, 3, 4, -12

# Generate x and y coordinates
x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)

X, Y = np.meshgrid(x, y)

# Calculate z using plane equation
Z = -(a*X + b*Y + d)/c

# Create 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot plane
ax.plot_surface(X, Y, Z, alpha=0.6)

# Calculate normal vector
normal = np.array([a, b, c])

# Plot normal vector from a point on the plane
P = np.array([0, 0, 3])

ax.quiver(
    P[0], P[1], P[2],
    normal[0], normal[1], normal[2],
    color='red',
    length=2,
    normalize=True,
    label='Normal Vector'
)

# Labels
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

ax.set_title('3D Plane and Its Normal Vector')

plt.show()
```

## Python code to normalize a 3D plane

```python
import numpy as np

# Plane equation: 2x + 3y + 6z - 12 = 0
plane = np.array([2, 3, 6, -12])

# Extract coefficients
a, b, c, d = plane

# Calculate normal vector
normal = np.array([a, b, c])

# Calculate magnitude
magnitude = np.linalg.norm(normal)

# Normalize plane
normalized_plane = plane / magnitude

# Extract normalized values
nx, ny, nz, d_normalized = normalized_plane

# Calculate distance from origin
distance = abs(d_normalized)

print("Original plane:", plane)

print("\nNormalized plane:")
print(normalized_plane)

print("\nNormalized normal vector:")
print([nx, ny, nz])

print("\nMagnitude of normal vector:")
print(np.linalg.norm([nx, ny, nz]))

print("\nDistance from origin:")
print(distance)
```

## Representing the normal vector using spherical coordinates

```python
import numpy as np

# Normalized normal vector
nx, ny, nz = 2/7, 3/7, 6/7

# Calculate spherical angles
theta = np.arctan2(ny, nx)
phi = np.arcsin(nz)

# Convert radians to degrees
theta_deg = np.degrees(theta)
phi_deg = np.degrees(phi)

print("Theta:", theta_deg, "degrees")
print("Phi:", phi_deg, "degrees")
```

##  Python code to visualize a 3D line using two points

```python
import numpy as np
import matplotlib.pyplot as plt

# Initial point
P = np.array([1, 2, 3])

# Direction vector
d = np.array([3, 4, 6])

# Normalize direction vector
d_hat = d / np.linalg.norm(d)

# Parameter
lam = np.linspace(-5, 5, 100)

# Calculate line points
R = P + np.outer(lam, d_hat)

# Extract coordinates
x = R[:, 0]
y = R[:, 1]
z = R[:, 2]

# Plot
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, label='3D Line')

# Plot initial point
ax.scatter(*P, color='red', s=100, label='Initial Point')

# Plot direction vector
ax.quiver(
    *P,
    *d_hat,
    length=2,
    color='green',
    label='Direction Vector'
)

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

ax.set_title('3D Line: Point-Direction Representation')
ax.legend()

plt.show()
```
